```python
#collect all mathematical constants
import cmath
def quadratic(a, b, c):
    discriminant = b ** 2 -(4*a*c)
    if discriminant < 0:
        print("No real roots")
        return
    else:
        sqrt_d = discriminant ** 0.5
        answer1 = (-b + sqrt_d)/(2*a)
        answer2 = (-b - sqrt_d)/(2*a)
        print(f"Answer1 = {answer1}")
        print(f"Answer2 = {answer2}")

def cubic(a,b ,c, d):
    #Compute variables within the core Formula
    delta_1 = (2*(b**3)) - 9*a*b*c + 27*(a**2)*d
    delta_0 = b**2 - 3*a*c
    discriminant = delta_1 **2 - 4 *(delta_0 **3)
    discriminant = cmath.sqrt(discriminant)
    inner_root = (delta_1 - discriminant)/2
    core_formula = inner_root ** (1/3)

    #Compute Omega
    omega = (-1 + cmath.sqrt(-3))/2
    #Compute Omega into root formula for k = 1, k = 2, k =3
    for k in range(0, 3):
        inner_bracket = b + ((omega ** k) * core_formula) + (delta_0) / ((omega ** k) * core_formula)
        x = (-1 / (3*a)) * inner_bracket
        print(x)


print("Quadratic, Cubic and Quartic Calculator")
user_input = input("Choose Your Equation Quadratic(1), Cubic(2), Quartic(3)")
if user_input == "1":
    a = float(input("A: "))
    b = float(input("B: "))
    c = float(input("C: "))
    quadratic(a, b, c)
elif user_input == "2":
    a = float(input("A: "))
    b = float(input("B: "))
    c = float(input("C: "))
    d = float(input("D: "))
    cubic(a ,b ,c ,d)
else:
    print("Equation Not available try again!")
    exit()
    


```
